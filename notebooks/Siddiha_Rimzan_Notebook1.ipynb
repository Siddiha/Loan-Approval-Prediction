Import Required Libraries

```python
# Import pandas for data manipulation
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Import plotly for interactive visualizations
import plotly.express as px

# Import LabelEncoder for categorical encoding
from sklearn.preprocessing import LabelEncoder

# Set pandas display options
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

Load the DataSet


```python
# Load the loan approval dataset
df = pd.read_csv('../data/loan_approval_data.csv')

print(f"Dataset loaded successfully!")
print(f"Shape: {df.shape}")
print(f"Number of rows: {df.shape[0]:,}")
print(f"Number of columns: {df.shape[1]}")
df.head()
```

Column Name Issues

```python
# Check if misspelled column exists
if 'emplyment_length' in df.columns:
    # Rename the misspelled column to correct spelling
    df.rename(columns={'emplyment_length': 'employment_length'}, inplace=True)
    # Print confirmation
    print("✓ Renamed 'emplyment_length' to 'employment_length'")
else:
    # If already correct, print message
    print("Column name already correct")
```

Initial Data Exploration

```python
# display first few rows
print("Frist 5 rows of the dataset")
df.head()

# Display column names and data types
print("Column Information:")
df.info()

# Display all column names
print("All column names:")
print(df.columns.tolist())
```

```python
# Check relationship between loan_approval_status and max_allowed_loan
print("Analyzing loan_approval_status and max_allowed_loan relationship:")
print("="*60)

# Group by loan_approval_status and show statistics for max_allowed_loan
print("\nStatistics by loan_approval_status:")
print(df.groupby('loan_approval_status')['max_allowed_loan'].describe())

# Check how many positive max_allowed_loan in each status
print("\nApproved loans (max_allowed_loan > 0) by status:")
print(df[df['max_allowed_loan'] > 0]['loan_approval_status'].value_counts())

# Check how many zero/negative max_allowed_loan in each status
print("\nRejected loans (max_allowed_loan <= 0) by status:")
print(df[df['max_allowed_loan'] <= 0]['loan_approval_status'].value_counts())
```

# DATASET 1 - CLASSIFICATION

## Task 1: Feature Selection for Classification


Dropping irrelevant features based on domain understanding


```python
# Create a copy of original data for classification
df_classification = df.copy()

# Drop 'id' column because it's a unique identifier with no predictive value
# Drop 'max_allowed_loan' column because it causes data leakage for classification
df_classification = df_classification.drop(columns=['id', 'max_allowed_loan'])

# Display number of retained features
print(f"Total retained features: {df_classification.shape[1]}")

# Display list of retained column names
print(f"\nRetained columns:")
print(df_classification.columns.tolist())
```

# Task 2: Data Exploration

Analyzing retained variables

```python
#Basic descriptive statistics for numerical variables
print("Descriptive Statistics for Numerical Variables: ")
#Calculate and display statistics (count, mean, std, min, quartiles, max)
df_classification.describe()
```

```python
# Display descriptive statistics for categorical variables
print("Descriptive Statistics - Categorical Variables:")
# Include only object type columns in describe
df_classification.describe(include='object')
```

```python
# Select columns with numerical data types (int64, float64)
numerical_vars = df_classification.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Select columns with categorical data type (object)
categorical_vars = df_classification.select_dtypes(include=['object']).columns.tolist()

# Display variable types
print("Variable Types:")
print(f"\nNumerical variables ({len(numerical_vars)}): {numerical_vars}")
print(f"\nCategorical variables ({len(categorical_vars)}): {categorical_vars}")
```

#Target Variable Distribution

```python
# Count occurrences of each value in target variable
print("Loan Approval Status Distribution:")
print(df_classification['loan_approval_status'].value_counts())

# Calculate and display percentage distribution
print("\nPercentage:")
# Normalize=True converts counts to proportions, multiply by 100 for percentage
print(df_classification['loan_approval_status'].value_counts(normalize=True) * 100)
```

# Task 3a: Identifying Data Quality issues

Check Missing Values

```python
# Count missing values
missing = df_classification.isnull().sum()

print("Missing Values:")
if missing.sum() == 0:
    print("✓ No missing values found!")
else:
    print(missing[missing > 0])
```

Check Duplicates

```python
# Check for duplicate rows
print("Duplicate Rows Analysis:")
print("="*60)
duplicates = df_classification.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
print(f"Percentage of duplicates: {(duplicates/len(df_classification))*100:.2f}%")
```

Check unrealistic Values

```python
#Check age values
print(f"Max age: {df_classification['age'].max()}")
unrealistic_age = len(df_classification[df_classification['age'] > 100])
print(f"Rows with age > 100: {unrealistic_age}")
```

Detect Outliers(IQR Method)

```python
# Detect outliers for each numerical variable
print("Outlier Detection (IQR Method):")
print("="*50)

for col in numerical_vars:
    if col != 'loan_approval_status':
        Q1 = df_classification[col].quantile(0.25)
        Q3 = df_classification[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df_classification[(df_classification[col] < lower) | (df_classification[col] > upper)]
        print(f"{col}: {len(outliers)} outliers ({len(outliers)/len(df_classification)*100:.1f}%)")

```

Visualize outliers before Treatment

```python
# Create box plots
plot_vars = ['age', 'income', 'employment_length', 'loan_amount', 'loan_interest_rate', 'loan_income_ratio']

df_melt_before = df_classification[plot_vars].melt(var_name='Variable', value_name='Value')

fig = px.box(
    df_melt_before,
    x='Variable',
    y='Value',
    title='Outliers BEFORE Treatment',
    color='Variable',
    color_discrete_sequence=['lightcoral'] * len(plot_vars),
    facet_col='Variable',
    facet_col_wrap=3
)
fig.update_layout(
    title_font_size=16,

    showlegend=False
)
fig.for_each_annotation(lambda a: a.update(text=a.text.split('=')[-1]))
fig.show()
```

#Task 3b: Fixing Data Quality Issues


Remove Duplicates

```python
# Remove duplicate rows
print(f"Before: {df_classification.shape[0]} rows")
df_classification = df_classification.drop_duplicates()
print(f"After: {df_classification.shape[0]} rows")
print(f"✓ Removed duplicates")
```

Cap Unrealistic Ages

```python

# Cap age at 100
unrealistic_count = len(df_classification[df_classification['age'] > 100])
if unrealistic_count > 0:
    df_classification['age'] = np.clip(df_classification['age'], 0, 100)
    print(f"✓ Capped {unrealistic_count} age values at 100")
else:
    print("✓ No unrealistic ages found")
```

Handle Outliers

```python
# Cap outliers using IQR method
print("Capping outliers...")
print("="*50)

for col in numerical_vars:
    if col != 'loan_approval_status':
        Q1 = df_classification[col].quantile(0.25)
        Q3 = df_classification[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        before = ((df_classification[col] < lower) | (df_classification[col] > upper)).sum()
        df_classification[col] = np.clip(df_classification[col], lower, upper)

        print(f"{col}: {before} outliers capped")
```

Visualize Outliers

```python
# Create box plots after treatment
df_melt_after = df_classification[plot_vars].melt(var_name='Variable', value_name='Value')

fig = px.box(
    df_melt_after,
    x='Variable',
    y='Value',
    title='Outliers AFTER Treatment',
    color='Variable',
    color_discrete_sequence=['lightgreen'] * len(plot_vars),
    facet_col='Variable',
    facet_col_wrap=3
)
fig.update_layout(
    title_font_size=14,
    showlegend=False
)
fig.for_each_annotation(lambda a: a.update(text=a.text.split('=')[-1]))
fig.show()
```

Encode Categorical Varaibles

```python
# Label encode binary variable
le = LabelEncoder()
df_classification['payment_default_on_file'] = le.fit_transform(df_classification['payment_default_on_file'])

print("Label Encoding:")
print(dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
# One-hot encode nominal variables
print(f"Before: {df_classification.shape}")

df_classification = pd.get_dummies(df_classification,
                                   columns=['home_ownership', 'loan_intent'],
                                   drop_first=True,
                                   dtype=int)

print(f"After: {df_classification.shape}")
print(f"\nColumns: {df_classification.columns.tolist()}")
```

Final Cleaned Dataset Summary

```python
# Display summary
print("DATASET 1 - CLASSIFICATION (Cleaned)")
print("="*50)
print(f"Shape: {df_classification.shape}")
print(f"Features: {df_classification.shape[1] - 1}")
print(f"Target: loan_approval_status")
print(f"Samples: {df_classification.shape[0]:,}")
print(f"Missing: {df_classification.isnull().sum().sum()}")
print(f"Duplicates: {df_classification.duplicated().sum()}")
```

Save Cleaned Dataset for Classification Modeling

```python
# Save to CSV
df_classification.to_csv('../data/dataset1_classification_cleaned.csv', index=False)
print("✓ Saved: dataset1_classification_cleaned.csv")
print(f"  Shape: {df_classification.shape}")
```

Check Missing Values

```python
# Check missing
missing_reg = df_regression.isnull().sum()
print("Missing Values:")
if missing_reg.sum() == 0:
    print("✓ No missing values!")
else:
    print(missing_reg[missing_reg > 0])
```

Removing Duplicates

```python
# Remove duplicates
print(f"Before: {df_regression.duplicated().sum()} duplicates")
df_regression = df_regression.drop_duplicates()
print(f"After: {df_regression.duplicated().sum()} duplicates")
print(f"Shape: {df_regression.shape}")
```

Cap Unrealistic Ages

```python
# Cap ages
unrealistic_reg = len(df_regression[df_regression['age'] > 100])
if unrealistic_reg > 0:
    df_regression['age'] = np.clip(df_regression['age'], 0, 100)
    print(f"✓ Capped {unrealistic_reg} ages at 100")
else:
    print("✓ No unrealistic ages")
```

Handle Outliers

```python
# Identify numerical variables
numerical_vars_reg = df_regression.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Cap outliers
print("Capping outliers (Regression):")
print("="*50)

for col in numerical_vars_reg:
    Q1 = df_regression[col].quantile(0.25)
    Q3 = df_regression[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    before = ((df_regression[col] < lower) | (df_regression[col] > upper)).sum()
    df_regression[col] = np.clip(df_regression[col], lower, upper)
    print(f"{col}: {before} outliers capped")
```

Encode categorical variables

```python
# Label encode
le_reg = LabelEncoder()
df_regression['payment_default_on_file'] = le_reg.fit_transform(df_regression['payment_default_on_file'])
print("Label Encoding:")
print(dict(zip(le_reg.classes_, le_reg.transform(le_reg.classes_))))
```

```python
# One-hot encode
print(f"Before: {df_regression.shape}")
df_regression = pd.get_dummies(df_regression,
                               columns=['home_ownership', 'loan_intent'],
                               drop_first=True,
                               dtype=int)
print(f"After: {df_regression.shape}")
print(f"\nColumns: {df_regression.columns.tolist()}")
```
